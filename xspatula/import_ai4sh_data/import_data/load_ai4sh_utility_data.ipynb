## Manage observation utilities

To run the next process you need to construct a [job_file][#] listing the JSON files creaeted in the first cell. The list of processes from the first cell shows all process files as absoulte paths, in the syntax of your operating system. To actually insert the imported data to your database, you need to follow these steps: 
1. create a directory where you want to save the list as a [pilot_file](#) (default: ./project/import_data/insert_data_pilots),
2. create and empty pilot text (.txt) file in that directorye (default: ai4sh_insert_utility_data.txt),
3. copy and paste the list output from the cell above to the pilot (.txt) file, and
4. create a [job_file](#) that points to the [pilot_file](#) (default: ./project/import_data/job_insert_tabular_utility_data.json),
5. if you changed the parent directory vis-a-vis the translation in the translation prython cell you must also to create a new [scheme_file](#) and reset the scheme file. If you kept your output from the first cell under the same project path as the translation you do not need to change scheme file.

If you used the default settings form the cloned (downloaded) GitHub repo, all you have to do is to paste the list in the (empty) pilot file already in place (./project/import_data/insert_data_pilots/ai4sh_insert_utility_data.txt).
